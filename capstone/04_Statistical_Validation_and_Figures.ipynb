```python
# ==============================================================================
# MODULE 4: STATISTICAL VALIDATION AND ODDS RATIOS
# Kāneʻohe Bay Coral Restoration Spatial Framework
# ==============================================================================

import os
import shutil
import numpy as np
import pandas as pd
from scipy.stats import chi2_contingency
from google.colab import drive

# --- PROACTIVE MOUNT POINT CLEANSE & SMART DRIVE MOUNT ---
mount_point = '/content/drive'
try:
    drive.flush_and_unmount()
except:
    pass

if os.path.exists(mount_point):
    try:
        shutil.rmtree(mount_point)
    except Exception as e:
        print(f"Clearing mountpoint warning: {e}")

os.makedirs(mount_point, exist_ok=True)
drive.mount(mount_point, force_remount=True)

CAPSTONE_ROOT = "/content/drive/MyDrive/Capstone"

# Chi-Square & Odds Ratio Validation (Benthic & Architecture Resilience)
# Contingency matrix comparing mid-century survival counts (Patchy vs. Dense)
contingency_patchy_vs_dense = np.array([[37218, 103760], [6622, 31881]])
chi2, p_val, dof, ex = chi2_contingency(contingency_patchy_vs_dense)
odds_ratio = (37218 * 31881) / (103760 * 6622)

print("--- STATISTICAL VALIDATION RESULTS ---")
print(f"Chi-Square Statistic: {chi2:.2f} (p-value: {p_val:.5e})")
print(f"Patchy vs Dense Coral Odds Ratio (OR): {odds_ratio:.2f}")
print("Statistical validation complete.")
```
